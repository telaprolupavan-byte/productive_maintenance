```python
def first_look(df, name):   # def = "define a recipe" called first_look
                            # (df, name) = two blanks to fill in each time you use it:
                            #    df   → which table to inspect
                            #    name → a label to print, like "failures"
                            # the colon means "the recipe's steps come next"

    print("=" * 60)         # "=" repeated 60 times: a divider line between tables
    print(name, df.shape)   # the label, then (rows, columns); shape is a value, so no ()
    print(df.head(3))       # the first 3 rows
    print(df.tail(3))       # the last 3 rows
    print()                 # an empty line for better readability between tables
```

```python
def first_look(df, name):
    """Print a quick health check of one table."""
    print("=" * 60)
    print(f"{name.upper()}   shape: {df.shape}")

    print("\n--- first and last rows ---")
    print(df.head(3))
    print(df.tail(3))

    print("\n--- column types ---")                       # step 3
    print(df.dtypes)

    print("\n--- missing values per column ---")          # step 4
    print(df.isna().sum())
    print("duplicate rows:", df.duplicated().sum())

    if "datetime" in df.columns:                           # step 5: only tables with dates
        print("\n--- time range ---")
        print(df["datetime"].min(), "→", df["datetime"].max())

    if "machineID" in df.columns:                          # step 6
        print("\nunique machines:", df["machineID"].nunique())

    if "machineID" in df.columns and "datetime" in df.columns:
        dup_keys = df.duplicated(subset=["machineID", "datetime"]).sum()
        print("same machine + same time rows:", dup_keys)

    text_cols = [c for c in df.columns                     # step 7: find the text columns
                 if pd.api.types.is_string_dtype(df[c])]
    for col in text_cols:
        print(f"\n--- value counts: {col} ---")
        print(df[col].value_counts())

    num = df.select_dtypes("number").drop(columns="machineID", errors="ignore")
    if num.shape[1] > 0:                                   # step 8: only if number columns remain
        print("\n--- number summary ---")
        print(num.describe().round(2))
```

```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data/raw")

FILES = {
    "telemetry": "PdM_telemetry.csv",
    "errors":    "PdM_errors.csv",
    "maint":     "PdM_maint.csv",
    "failures":  "PdM_failures.csv",
    "machines":  "PdM_machines.csv",
}

tables = {}
for name, file in FILES.items():
    df = pd.read_csv(DATA_DIR / file)

    if "datetime" in df.columns:
        df["datetime"] = pd.to_datetime(df["datetime"])   # only for tables that have dates
        date_info = str(df["datetime"].dtype)
    else:
        date_info = "no datetime column"                    # machines lands here

    tables[name] = df                                       # store every table, dates or not
    print(f"{name:<10} {str(df.shape):<14} {date_info}")
```
